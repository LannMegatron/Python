```python
import numpy as np
import matplotlib.pyplot as plt

xstart = 0
xstop = 2*np.pi
increment = 0.1

x = np.arange(xstart, xstop, increment)
y = np.cos(x)

plt.plot(x,y)
plt.title('Grafik cos 2pi')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

# LATIHAN 3

```python
import numpy as np
import matplotlib.pyplot as plt

xstart = 0
xstop = 4*np.pi
increment = 0.1

x = np.arange(xstart, xstop, increment)
y = np.sin(x)
z = np.cos(x)

plt.subplot(2,1,1)
plt.plot(x,y,'g')
plt.title('Grafik Sin 4pi')
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.grid()
plt.show()

plt.subplot(2,1,2)
plt.plot(x,z,'r')
plt.title('Grafik Cos 4pi')
plt.xlabel('x')
plt.ylabel('cos(x)')
plt.grid()
plt.show()
```

# LATIHAN 4

```python
import math as mt
import numpy as np
import matplotlib.pyplot as plt

T = 10
a = -1/T

x0 = 5
t = 0

tstart = 0
tstop = 50
increment = 1

x = []
x = np.zeros(tstop+1)
t = np.arange(tstart, tstop+1, increment)

for k in range (tstop):
    x[k] = mt.exp(a*t[k])*x0


plt.plot(t,x)
plt.title('Simulation of Dynamic System')
plt.xlabel('x')
plt.ylabel('y')
plt.grid()
plt.axis([0, 50, 0, 5])
plt.show()
```
